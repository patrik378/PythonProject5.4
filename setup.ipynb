```python
a = int(input())

for i in range(1, a+1):
    for j in range(1, a+1):
        if i == j:
            print(1, end=' ')
        else:
            print(0, end=' ')
    print()
```

```python
a = int(input())
b = int(input())
PARI = 0

for x in range(1, a+1):
    for y in range(1, b+1):

        if (x + y) % 2 == 0:
            PARI += 1

print(PARI)
```
